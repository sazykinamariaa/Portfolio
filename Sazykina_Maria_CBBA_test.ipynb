Названия категорий в budget не совпадают с названиями проф.групп в hr (например, «Инженерия ПО» в budget = «Разработка» в hr), нужно будет вручную сопоставить категории.

```python
# Сопоставляем категории бюджета с проф.группами из hr
budget_mapping = {'Инженерия ПО': 'Разработка',
    'Дата-аналитика': 'Аналитика данных',
    'Эксплуатация и DevOps': 'Инфраструктура',
    'Финансовое планирование': 'Бюджетирование',
    'C&B': 'Компенсации и льготы',
    'Персонал и найм': 'HR',
    'Клиентский сервис': 'Поддержка'}

print(pd.DataFrame(list(budget_mapping.items()), columns=['Категория бюджета', 'Проф. группа в hr']))
```

Названия категорий в таблице бюджета и названия проф.групп в hr написаны по-разному, но означают одно и то же (например, «Инженерия ПО» = «Разработка»). Таблица выше показывает кто кому соответствует - это понадобится когда будем распределять бюджет на повышения.

```python
# Смотрим уникальные значения rate, salary и grade, чтобы понять, что не так с форматом
print(hr['rate'].unique())
print(hr['salary'].unique())
print(hr['grade'].unique())
```

Что заметно после **unique()**:

**rate**: одно и то же значение записано в разных форматах — десятичная дробь с точкой (1.0), с запятой (1,0) и в процентах (100%). Для расчётов нужно привести всё к единому числовому формату от 0 до 1.

**salary**: пробелы внутри числа: 118 500, 71 000; слово «руб»: 135000 руб, 185000 руб; комбинация обоих: 85200 руб, 118500 руб. Необходимо это убрать и перевести в int.

**grade**: присутсвует опечатка в виде "118" и nan. Нужно убрать опечатку и посмотреть, что за недостающие значения

```python
# Чистим salary: убираем слово "руб" и пробелы, переводим в число
hr['salary'] = hr['salary'].str.replace('руб', '', regex=False)
hr['salary'] = hr['salary'].str.replace(' ', '', regex=False)
hr['salary'] = hr['salary'].astype(int)

# Проверяем результат
print(hr['salary'].unique())
```

```python
# Чистим rate: убираем % и заменяем запятые на точки, переводим в число
hr['rate'] = hr['rate'].str.replace('%', '', regex=False)
hr['rate'] = hr['rate'].str.replace(',', '.', regex=False)
hr['rate'] = hr['rate'].astype(float)

# Если значение больше 1, то это проценты, делим на 100
hr['rate'] = hr['rate'].apply(lambda x: x / 100 if x > 1 else x)

# Проверяем результат
print(hr['rate'].unique())
```

```python
# Находим строки с грейдом 118
print(hr[hr['grade'] == 118])

# Исправляем 118 на 18
hr['grade'] = hr['grade'].replace(118, 18)
print(hr['grade'].unique())
```

```python
# Находим сотрудников с пустым грейдом
print(hr[hr['grade'].isna()])

# Смотрим вилки для их проф.групп чтобы определить грейд вручную
print(bands[bands['prof_group'].isin(['Разработка', 'Бюджетирование'])])

# Заполняем грейды вручную на основе salary_bands:
# Роман Иванов (разработка, зарплата 209000) — попадает в grade 16 (min 178000, mid 210000, max 241000)
# ближе к mid (отклонение 1000) чем grade 15 (mid 200000, отклонение 9000)
hr.loc[hr['employee_id'] == 1004, 'grade'] = 16

# Григорий Васильев (бюджетирование, зарплата 175000) — попадает в grade 15 (min 143000, mid 168000, max 193000)
hr.loc[hr['employee_id'] == 1050, 'grade'] = 15

# Проверка
print(hr['grade'].isna().sum())
```

```python
# Переводим grade в int
hr['grade'] = hr['grade'].astype(int)
print(hr['grade'].unique())
```

```python
# Проверим уник знач prof_group
print(hr['prof_group'].unique())
```

**prof_group** - значения одной группы записаны по-разному:

1) Разработка, разработка, Разраб., Разработка  (с пробелом)

2) Аналитика данных, аналитика данных, Аналитика данных  (с пробелом)

3) Компенсации и льготы, Компенсации и льготы  (с пробелом), Компенсации, C&B

Необходимо привести к единому стандарту.

```python
# Приводим prof_group к единому формату, убираем пробелы по краям
hr['prof_group'] = hr['prof_group'].str.strip()

# Стандартизируем
hr['prof_group'] = hr['prof_group'].replace({
    'разработка': 'Разработка',
    'Разраб.': 'Разработка',
    'аналитика данных': 'Аналитика данных',
    'Компенсации': 'Компенсации и льготы',
    'C&B': 'Компенсации и льготы'
})

print(hr['prof_group'].unique())
```

```python
# Проверим уник знач hire_date
print(hr['hire_date'].unique())
```

**hire_date** — два формата дат: 2022-03-01 (YYYY-MM-DD) и 01.03.2022 (DD.MM.YYYY), приводим к единому YYYY-MM-DD.

```python
# Единый формат даты
hr['hire_date'] = pd.to_datetime(hr['hire_date'], format='mixed', dayfirst=True)
hr['hire_date'] = hr['hire_date'].dt.strftime('%Y-%m-%d')

print(hr['hire_date'].unique())
```
